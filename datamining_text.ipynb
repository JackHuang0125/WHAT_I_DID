```python
import nltk
from nltk.corpus import gutenberg
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

# 確保下載需要的語料庫
nltk.download('gutenberg')

# Step 1: 選擇文本 (例如：Pride and Prejudice)
text = gutenberg.raw('austen-emma.txt')  # 以 Jane Austen 的 Emma 為例

# Step 2: 初始化 CountVectorizer
vectorizer = CountVectorizer()

# Step 3: 進行詞頻矩陣計算
X = vectorizer.fit_transform([text])  # 需要將文本包裝成列表以適配 CountVectorizer 的輸入

# Step 4: 取得詞表和頻率
vocabulary = vectorizer.get_feature_names_out()  # 詞表
word_counts = X.toarray()[0]  # 詞頻數值

# Step 5: 建立 DataFrame 便於查看
df = pd.DataFrame({'word': vocabulary, 'count': word_counts})

# Step 6: 排序詞頻並查看最常見的詞
df = df.sort_values(by='count', ascending=False)
```

```python
import requests
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import pandas as pd

# 確保下載必要的資源
import nltk
nltk.download('punkt')
nltk.download('stopwords')

# 停用詞列表
stop_words = set(stopwords.words('english'))

# Gutenberg ID 列表
gutenberg_ids = [1260, 768, 969, 9182, 767]

# 儲存所有文本
all_texts = ""

# 從 Project Gutenberg 網站下載文本
for book_id in gutenberg_ids:
    url = f"https://www.gutenberg.org/files/{book_id}/{book_id}-0.txt"
    response = requests.get(url)
    
    if response.status_code == 200:
        print(f"Successfully downloaded book ID {book_id}")
        text = response.text  # 取得文本內容
        all_texts += text  # 將文本拼接
    else:
        print(f"Failed to download book ID {book_id}")

# 分詞
tokens = word_tokenize(all_texts.lower())

# 移除停用詞與非單詞字符
filtered_tokens = [word for word in tokens if word.isalnum() and word not in stop_words]

# 詞頻統計
word_counts = pd.Series(filtered_tokens).value_counts()

word_counts
```
