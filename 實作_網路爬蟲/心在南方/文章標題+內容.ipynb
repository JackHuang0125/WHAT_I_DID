```python
import pandas as pd
from bs4 import BeautifulSoup as B
import requests as req
import csv
import time
import random

headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:137.0) Gecko/20100101 Firefox/137.0'}
titles = []
urls = []

for i in range(1, 73):
    print(f'正在處理第{i}頁')
    base_url = f'https://stat.nuk.edu.tw/South.asp?page={i}&mysort='
    resp = req.get(base_url, verify=False ,headers=headers)
    resp.encoding = 'utf-8'
    soup = B(resp.text, 'html.parser')

    time.sleep(random.uniform(0.5, 1.5))

    for row in soup.find_all('tr'):
        authors = row.find('td', class_="zt14px1")
        if authors and any(name in authors.text for name in ['黃文璋', '南方壺', '學易']):
            spans = row.find('span', class_='font_blue')
            a_tag = spans.find('a') if spans else None
            if a_tag and a_tag.has_attr('href'):
                title = spans.text.strip()
                titles.append(title)
                inner_url = 'https://stat.nuk.edu.tw/'+ a_tag['href']
                urls.append(inner_url)

df = pd.DataFrame({'title': titles, 'url': urls})
df = df.drop_duplicates(subset='title').reset_index(drop=True)  # 可選：移除重複

test = []

for u in df['url']:
    resp = req.get(u, verify=False)
    resp.encoding = 'utf-8'
    soup = B(resp.text, 'html.parser')

    for tds in soup.find_all('td', align="left", valign="top"):
        for td in tds:
            data = td.get_text(strip=True)
            if data:
                test.append(data)
                
    time.sleep(random.uniform(1.0, 3.0))

df['data'] = test
df.head()
```

```python
import pandas as pd
from bs4 import BeautifulSoup as B
import requests as req
import csv
import time
import random

headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:137.0) Gecko/20100101 Firefox/137.0'}
titles = []
urls = []
authors = []
dates = []

for i in range(1, 73):
    print(f'正在處理第{i}頁')
    base_url = f'https://stat.nuk.edu.tw/South.asp?page={i}&mysort='
    resp = req.get(base_url, verify=False ,headers=headers)
    resp.encoding = 'utf-8'
    soup = B(resp.text, 'html.parser')

    time.sleep(random.uniform(0.5, 1.5))

    for row in soup.find_all('tr'):
        author = row.find('td', class_="zt14px1", width="19%")
        if author and any(name in author.text for name in ['黃文璋', '南方壺', '學易']):
            authors.append(author.text.strip())
            spans = row.find('span', class_='font_blue')
            times = row.find('td', class_='english')

            titles.append(spans.text.strip())
            dates.append(times.text.strip())
                

new_df = pd.DataFrame({'title': titles, 'author': authors, 'date': dates})
new_df = new_df.drop_duplicates(subset='title').reset_index(drop=True) 
```
