```python
import numpy as np

def estimate_arl1(mu, sigma, k=0.5, h=4.095, M=10**5, seed=611211106):
    rng = np.random.default_rng(seed)
    RL = np.zeros(M, dtype=int)

    for m in range(M):
        C_plus = 0
        n = 0

        while True:
            n += 1
            x = rng.normal(mu, sigma)

            C_plus = max(0, C_plus + x - k)

            if C_plus > h:
                RL[m] = n
                break

    return RL.mean()

def estimate_arl1_down(mu, sigma, k=0.5, h=4.095, M=10**5, seed=611211106):
    rng = np.random.default_rng(seed)
    RL = np.zeros(M, dtype=int)

    for m in range(M):
        C_minus = 0
        n = 0

        while True:
            n += 1
            x = rng.normal(mu, sigma)

            C_minus = max(0, C_minus - x - k)

            if C_minus > h:
                RL[m] = n
                break

    return RL.mean()

settings = [
    # (0.5, 1),
    # (1.0, 1),
    (-1.0, 1),
    (0.0, 2),
    (0.0, 0.5),
    (1.0, 2)
]

for mu, sigma in settings:
    if mu < 0:
        arl1 = estimate_arl1_down(mu, sigma)
    else:
        arl1 = estimate_arl1(mu, sigma)

    print(f"N({mu}, {sigma}^2), ARL1 = {arl1:.3f}")
```

```python
def estimate_arl1(mu, sigma, k=0.5, h=4.095, M=10**5, seed=611211106):
    rng = np.random.default_rng(seed)
    RL = np.zeros(M, dtype=int)

    max_steps = 100000

    for m in range(M):
        C_plus = 0
        n = 0

        while n < max_steps:
            n += 1
            x = rng.normal(mu, sigma)

            C_plus = max(0, C_plus + x - k)

            if C_plus > h:
                RL[m] = n
                break

        if RL[m] == 0:
            RL[m] = max_steps

    return RL.mean()

settings = [
    (0.0, 0.5),
    (1.0, 2)
]

for mu, sigma in settings:
    if mu < 0:
        arl1 = estimate_arl1_down(mu, sigma)
    else:
        arl1 = estimate_arl1(mu, sigma)

    print(f"N({mu}, {sigma}^2), ARL1 = {arl1:.3f}")
```
