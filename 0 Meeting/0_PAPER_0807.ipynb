```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, classification_report

df = banknote_merged.copy()
y_col, c_col = 'class', 'gap_cluster'
X_cols = [c for c in df.columns if c not in (y_col, c_col)]

# 1) 全域一次切分，對 (y, cluster) 聯合分層，確保各群的類別分布一致
strat_key = df[y_col].astype(str) + '|' + df[c_col].astype(str)
train_idx, test_idx = train_test_split(df.index, test_size=0.3, stratify=strat_key, random_state=42)
tr, te = df.loc[train_idx], df.loc[test_idx]

X_train, y_train, c_train = tr[X_cols], tr[y_col], tr[c_col]
X_test,  y_test,  c_test  = te[X_cols], te[y_col], te[c_col]

# 2) 全資料「打亂 y_train」的全域模型
rng = np.random.default_rng(42)
permutation = rng.permutation(len(y_train))
train_ids = y_train.index.to_numpy()
permutaion_ids = train_ids[permutation]
y_train_permutation = pd.Series(y_train.loc[permutaion_ids].to_numpy(), index=train_ids)

# 3) 全域模型（用打亂後的 y 訓練）
global_rfc = RandomForestClassifier(n_estimators=300, random_state=42, n_jobs=-1, class_weight='balanced')
global_rfc.fit(X_train, y_train_permutation.to_numpy())
pred_global = pd.Series(global_rfc.predict(X_test), index=X_test.index)
acc_global = accuracy_score(y_test, pred_global)

# 4) 逐 cluster 建模（同樣用打亂後的 y）
cluster_models = {}

rows = []
reports = {}
for k in sorted(tr[c_col].unique()):
    train_k = tr[c_train == k]
    test_k  = te[c_test == k]
    if len(test_k) == 0:
        continue

    Xtrain_k, ytrain_k = train_k[X_cols], train_k[y_col]
    Xtest_k,  ytest_k  = test_k[X_cols],  test_k[y_col]
    ytr_k_perm = y_train_permutation.loc[Xtrain_k.index].to_numpy()

    # 若打亂後訓練 y 只剩一類或樣本太少，改用 Dummy
    if (len(train_k) < 10) or (pd.Series(ytr_k_perm).nunique() < 2):
        c_model = DummyClassifier(strategy='most_frequent', random_state=42)
    else:
        c_model = RandomForestClassifier(n_estimators=300, class_weight='balanced',
                                         random_state=42, n_jobs=-1)
    c_model.fit(Xtrain_k, ytr_k_perm)
    cluster_models[k] = c_model

    # 同一群上的公平比較
    mask_k = (c_test == k)
    acc_g = accuracy_score(y_test[mask_k], pred_global[mask_k])
    f1_g  = f1_score(y_test[mask_k], pred_global[mask_k], average='macro', zero_division=0)

    pred_k = c_model.predict(Xtest_k)
    acc_k  = accuracy_score(ytest_k, pred_k)
    f1_k   = f1_score(ytest_k, pred_k, average='macro', zero_division=0)

    rows.append({
        'cluster': k,
        'n_train': len(train_k),
        'n_test': len(test_k),
        'unchanged_train': (ytrain_k.to_numpy() == ytr_k_perm).mean(),
        'Global_Acc': acc_g, 'Global_MacroF1': f1_g,
        'Cluster_Acc': acc_k, 'Cluster_MacroF1': f1_k,
        'Delta_Acc': acc_k - acc_g, 'Delta_MacroF1': f1_k - f1_g
    })

    reports[k] = classification_report(ytest_k, pred_k, zero_division=0, output_dict=False)

# 5) 彙整表與整體（全測試集）表現
summary = pd.DataFrame(rows).sort_values('cluster').reset_index(drop=True)
overall_acc_g = accuracy_score(y_test, pred_global)
overall_f1_g  = f1_score(y_test, pred_global, average='macro', zero_division=0)

print("=== Per-cluster comparison ===")
print(summary)

print("\n=== Global model on all test ===")
print(f"Acc={overall_acc_g:.3f}  MacroF1={overall_f1_g:.3f}")

print("\n=== Classification reports per cluster (cluster model) ===")
for k in reports:
    print(f"\n[cluster {k}]")
    print(reports[k])
```

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, classification_report

df = banknote_merged.copy()
y_col, c_col = 'class', 'gap_cluster'
X_cols = [c for c in df.columns if c not in (y_col, c_col)]

# 1) 全域一次切分，對 (y, cluster) 聯合分層，確保各群的類別分布一致
strat_key = df[y_col].astype(str) + '|' + df[c_col].astype(str)
train_idx, test_idx = train_test_split(df.index, test_size=0.3, stratify=strat_key, random_state=42)
tr, te = df.loc[train_idx], df.loc[test_idx]

X_train, y_train, c_train = tr[X_cols], tr[y_col], tr[c_col]
X_test,  y_test,  c_test  = te[X_cols], te[y_col], te[c_col]

# 2) 全資料「打亂 y_train」的全域模型
rng = np.random.default_rng(42)
permutation = rng.permutation(len(y_train))
train_ids = y_train.index.to_numpy()
permutaion_ids = train_ids[permutation]
y_train_permutation = pd.Series(y_train.loc[permutaion_ids].to_numpy(), index=train_ids)

# 3) 全域模型（用打亂後的 y 訓練）
global_rfc = RandomForestClassifier(n_estimators=300, random_state=42, n_jobs=-1, class_weight='balanced')
global_rfc.fit(X_train, y_train_permutation.to_numpy())
pred_global = pd.Series(global_rfc.predict(X_test), index=X_test.index)
acc_global = accuracy_score(y_test, pred_global)

# 4) 逐 cluster 建模（同樣用打亂後的 y）
cluster_models = {}

rows = []
reports = {}
for k in sorted(tr[c_col].unique()):
    train_k = tr[c_train == k]
    test_k  = te[c_test == k]
    if len(test_k) == 0:
        continue

    Xtrain_k, ytrain_k = train_k[X_cols], train_k[y_col]
    Xtest_k,  ytest_k  = test_k[X_cols],  test_k[y_col]
    ytr_k_perm = y_train_permutation.loc[Xtrain_k.index].to_numpy()

    # 若打亂後訓練 y 只剩一類或樣本太少，改用 Dummy
    if (len(train_k) < 10) or (pd.Series(ytr_k_perm).nunique() < 2):
        c_model = DummyClassifier(strategy='most_frequent', random_state=42)
    else:
        c_model = RandomForestClassifier(n_estimators=300, class_weight='balanced',
                                         random_state=42, n_jobs=-1)
    c_model.fit(Xtrain_k, ytr_k_perm)
    cluster_models[k] = c_model

    # 同一群上的公平比較
    mask_k = (c_test == k)
    acc_g = accuracy_score(y_test[mask_k], pred_global[mask_k])
    f1_g  = f1_score(y_test[mask_k], pred_global[mask_k], average='macro', zero_division=0)

    pred_k = c_model.predict(Xtest_k)
    acc_k  = accuracy_score(ytest_k, pred_k)
    f1_k   = f1_score(ytest_k, pred_k, average='macro', zero_division=0)

    rows.append({
        'cluster': k,
        'n_train': len(train_k),
        'n_test': len(test_k),
        'unchanged_train': (ytrain_k.to_numpy() == ytr_k_perm).mean(),
        'Global_Acc': acc_g, 'Global_MacroF1': f1_g,
        'Cluster_Acc': acc_k, 'Cluster_MacroF1': f1_k,
        'Delta_Acc': acc_k - acc_g, 'Delta_MacroF1': f1_k - f1_g
    })

    reports[k] = classification_report(ytest_k, pred_k, zero_division=0, output_dict=False)

# 5) 彙整表與整體（全測試集）表現
summary = pd.DataFrame(rows).sort_values('cluster').reset_index(drop=True)
overall_acc_g = accuracy_score(y_test, pred_global)
overall_f1_g  = f1_score(y_test, pred_global, average='macro', zero_division=0)

print("=== Per-cluster comparison ===")
print(summary)

print("\n=== Global model on all test ===")
print(f"Acc={overall_acc_g:.3f}  MacroF1={overall_f1_g:.3f}")

print("\n=== Classification reports per cluster (cluster model) ===")
for k in reports:
    print(f"\n[cluster {k}]")
    print(reports[k])
```

```python
import numpy as np

def route_and_predict(x_new, method='corr', t=0.6):
    router = router_corr if method=='corr' else router_euclid
    Xn = x_new[X_cols]

    # 取得路由機率，挑 Top-2 群
    P = router.predict_proba(Xn)[0]
    labels = router.classes_
    top2 = np.argsort(P)[-2:][::-1]
    k1, k2 = int(labels[top2[0]]), int(labels[top2[1]])
    p1, p2 = float(P[top2[0]]), float(P[top2[1]])

    # 低信心：退回全域
    if p1 < t:
        y_hat = global_rfc.predict(Xn)[0]
        return {'route':'global', 'y_pred':int(y_hat), 'p_router':p1, 'k1':k1, 'k2':k2}

    # Top-2 對決：選模型自身給的最大機率較大的那個
    proba1 = cluster_models[k1].predict_proba(Xn)[0].max()
    proba2 = cluster_models[k2].predict_proba(Xn)[0].max()
    if proba1 >= proba2:
        y_hat = cluster_models[k1].predict(Xn)[0]
        return {'route':k1, 'y_pred':int(y_hat), 'p_router':p1}
    else:
        y_hat = cluster_models[k2].predict(Xn)[0]
        return {'route':k2, 'y_pred':int(y_hat), 'p_router':p2}


# === 範例 ===
x_new = pd.DataFrame([{'variance': 1, 'skewness': 8.5, 'curtosis': -2.0, 'entropy': -1.5}])[X_cols]

print(route_and_predict(x_new, method='corr', t=0.6))  # 或 method='euclid'

# 看路由信心：kNN 機率或最近鄰距離
proba = router_corr.predict_proba(x_new[X_cols]).max(axis=1)[0]
dists, idx = router_corr.named_steps['kneighborsclassifier'].kneighbors(
    router_corr.named_steps['standardscaler'].transform(x_new[X_cols]),
    n_neighbors=11, return_distance=True)
print("route confidence =", proba, "  nearest-dist avg =", dists.mean())
```
