```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

df = banknote_merged.copy()
y_col, c_col = 'class', 'gap_cluster'
X_cols = [c for c in df.columns if c not in (y_col, c_col)]

# 1) 全域一次切分，對 (y, cluster) 聯合分層，確保各群的類別分布一致
strat_key = df[y_col].astype(str) + '|' + df[c_col].astype(str)
train_idx, test_idx = train_test_split(df.index, test_size=0.3, stratify=strat_key, random_state=42)
tr, te = df.loc[train_idx], df.loc[test_idx]

X_train, y_train, c_train = tr[X_cols], tr[y_col], tr[c_col]
X_test,  y_test,  c_test  = te[X_cols], te[y_col], te[c_col]

# 2) 全資料「打亂 y_train」的全域模型
rng = np.random.default_rng(42)
y_train_perm = y_train.to_numpy().copy()
rng.shuffle(y_train_perm)

global_perm = RandomForestClassifier(n_estimators=300, random_state=42, n_jobs=-1, class_weight='balanced')
global_perm.fit(X_train, y_train_perm)

pred_global = global_perm.predict(X_test)
acc_global = accuracy_score(y_test, pred_global)

# 3) 各 cluster 「打亂 y_train」模型（只在該群的訓練子集上學，測該群 test）
rows = []
cluster_models = {}
for k in sorted(c_train.unique()):
    tr_k = tr[c_train == k]
    te_k = te[c_test == k]

    # 若群太小或訓練只剩單一類別，改用 Dummy 以避免 RF 報錯
    Xtr_k, ytr_k = tr_k[X_cols], tr_k[y_col]
    Xte_k, yte_k = te_k[X_cols], te_k[y_col]

    if (len(tr_k) < 10) or (ytr_k.nunique() < 2):
        clf_k = DummyClassifier(strategy='most_frequent', random_state=42)
        ytr_perm_k = ytr_k  # Dummy 不用真的打亂
    else:
        clf_k = RandomForestClassifier(
            n_estimators=300, random_state=42, n_jobs=-1, class_weight='balanced'
        )
        rng_k = np.random.default_rng(42 + int(k))
        ytr_perm_k = ytr_k.to_numpy().copy()
        rng_k.shuffle(ytr_perm_k)

    clf_k.fit(Xtr_k, ytr_perm_k)
    cluster_models[k] = clf_k
    pred_k = clf_k.predict(Xte_k)
    acc_k = accuracy_score(yte_k, pred_k)

    # 全域模型在該群 test 的成績（逐群對照）
    # 建全域預測
    pred_global = global_perm.predict(X_test)

    # ✅ 對齊索引：用 X_test.index 當索引
    pred_global_s = pd.Series(pred_global, index=X_test.index)

    # 迴圈內取各群時改這行：
    pred_global_k = pred_global_s.loc[te_k.index].to_numpy()

    acc_global_k = accuracy_score(yte_k, pred_global_k)

    rows.append({
        'cluster': int(k),
        'n_train': len(tr_k),
        'n_test': len(te_k),
        'acc_cluster_perm': acc_k,
        'acc_global_perm_on_cluster': acc_global_k,
        'delta': acc_k - acc_global_k
    })

res = pd.DataFrame(rows).sort_values('cluster')

# 4) 聚合：Micro / Macro
micro_acc_clusters = np.average(res['acc_cluster_perm'], weights=res['n_test'])
macro_acc_clusters = res['acc_cluster_perm'].mean()

micro_acc_global_by_cluster = np.average(res['acc_global_perm_on_cluster'], weights=res['n_test'])
macro_acc_global_by_cluster = res['acc_global_perm_on_cluster'].mean()

print("=== Per-cluster (permuted-y) comparison ===")
print(res)

print("\n=== Aggregated accuracy ===")
print(f"Global permuted-y (overall test): {acc_global:.4f}")
print(f"Clusters permuted-y  Micro-Acc  : {micro_acc_clusters:.4f}")
print(f"Clusters permuted-y  Macro-Acc  : {macro_acc_clusters:.4f}")
print(f"Global permuted-y by-cluster Micro-Acc: {micro_acc_global_by_cluster:.4f}")
print(f"Global permuted-y by-cluster Macro-Acc: {macro_acc_global_by_cluster:.4f}")
print(f"Δ (Clusters Micro - Global overall): {(micro_acc_clusters - acc_global):.4f}")
```
