```python
import numpy as np, pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, f1_score, classification_report

df = banknote_merged.copy()
y_col, c_col = 'class', 'gap_cluster'      # c_col 來自「全資料 GAP」的群
X_cols = [c for c in df.columns if c not in (y_col, c_col)]

# 1) 只用 y 分層切分（避免洩漏）
train_idx, test_idx = train_test_split(df.index, test_size=0.3, stratify=df[y_col], random_state=42)
tr, te = df.loc[train_idx], df.loc[test_idx]
X_train, y_train, c_train = tr[X_cols], tr[y_col], tr[c_col]
X_test,  y_test,  c_test  = te[X_cols], te[y_col], te[c_col]

# 2) 固定「同一組打亂 y_train」
rng = np.random.default_rng(42)
perm = rng.permutation(len(y_train))
y_train_perm = pd.Series(y_train.iloc[perm].to_numpy(), index=y_train.index)

# 3) 全域模型（baseline）
global_rfc = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1, class_weight='balanced')
global_rfc.fit(X_train, y_train_perm.to_numpy())
pred_global = pd.Series(global_rfc.predict(X_test), index=X_test.index)
acc_global  = accuracy_score(y_test, pred_global)
f1_global   = f1_score(y_test, pred_global, average='macro', zero_division=0)

# 4) 逐群模型（用 c_train_hat / c_test_hat）
rows, reports, cluster_models = [], {}, {}
for k in sorted(c_train.unique()):
    idx_tr = (c_train == k)
    idx_te = (c_test == k)
    if idx_te.sum() == 0: 
        continue

    Xtr_k = X_train[idx_tr]; Xte_k = X_test[idx_te]
    ytr_k_perm = y_train_perm.loc[Xtr_k.index].to_numpy()
    yte_k = y_test[idx_te]

    if (idx_tr.sum() < 10) or (pd.Series(ytr_k_perm).nunique() < 2):
        model_k = DummyClassifier(strategy='most_frequent', random_state=42)
    else:
        model_k = RandomForestClassifier(n_estimators=100, class_weight='balanced', random_state=42, n_jobs=-1)
    model_k.fit(Xtr_k, ytr_k_perm)
    cluster_models[k] = model_k
    
    pk = model_k.predict(Xte_k)

    # 同群內（公平）比較：global vs cluster-k
    acc_g_k = accuracy_score(yte_k, pred_global[idx_te])
    f1_g_k  = f1_score(yte_k, pred_global[idx_te], average='macro', zero_division=0)
    acc_k   = accuracy_score(yte_k, pk)
    f1_k    = f1_score(yte_k, pk, average='macro', zero_division=0)

    # 該群 train 的「unchanged」比例（對齊 index 後比較）
    idx_tr_k = Xtr_k.index
    unchanged_frac_k = (y_train.loc[idx_tr_k] == y_train_perm.loc[idx_tr_k]).mean()
    
    rows.append({'cluster': int(k), 'n_train': int(idx_tr.sum()), 'n_test': int(idx_te.sum()), 'unchanged_train': unchanged_frac_k,
                 'Global_Acc': acc_g_k, 'Global_MacroF1': f1_g_k,
                 'Cluster_Acc': acc_k, 'Cluster_MacroF1': f1_k,
                 'Delta_Acc': acc_k-acc_g_k, 'Delta_MacroF1': f1_k-f1_g_k})
    reports[k] = classification_report(yte_k, pk, zero_division=0)
    
# 6) 彙整
summary = pd.DataFrame(rows).sort_values('cluster').reset_index(drop=True)

print("=== Global model on all test ===")
print(f"Acc={acc_global:.3f}  MacroF1={f1_global:.3f}")
print("\n=== Cluster comparison ===")
print(summary)
```

```python
import numpy as np, pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, f1_score, classification_report

df = banknote_merged.copy()
y_col, c_col = 'class', 'gap_cluster'      # c_col 來自「全資料 GAP」的群
X_cols = [c for c in df.columns if c not in (y_col, c_col)]

# 1) 只用 y 分層切分（避免洩漏）
train_idx, test_idx = train_test_split(df.index, test_size=0.3, stratify=df[y_col], random_state=42)
tr, te = df.loc[train_idx], df.loc[test_idx]
X_train, y_train, c_train = tr[X_cols], tr[y_col], tr[c_col]
X_test,  y_test,  c_test  = te[X_cols], te[y_col], te[c_col]

# 2) 固定「同一組打亂 y_train」
rng = np.random.default_rng(42)
perm = rng.permutation(len(y_train))
y_train_perm = pd.Series(y_train.iloc[perm].to_numpy(), index=y_train.index)

# 3) 全域模型（baseline）
global_rfc = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1, class_weight='balanced')
global_rfc.fit(X_train, y_train_perm.to_numpy())
pred_global = pd.Series(global_rfc.predict(X_test), index=X_test.index)
acc_global  = accuracy_score(y_test, pred_global)
f1_global   = f1_score(y_test, pred_global, average='macro', zero_division=0)

# 4) 逐群模型（用 c_train_hat / c_test_hat）
rows, reports, cluster_models = [], {}, {}
for k in sorted(c_train.unique()):
    idx_tr = (c_train == k)
    idx_te = (c_test == k)
    if idx_te.sum() == 0: 
        continue

    Xtr_k = X_train[idx_tr]; Xte_k = X_test[idx_te]
    ytr_k_perm = y_train_perm.loc[Xtr_k.index].to_numpy()
    yte_k = y_test[idx_te]

    if (idx_tr.sum() < 10) or (pd.Series(ytr_k_perm).nunique() < 2):
        model_k = DummyClassifier(strategy='most_frequent', random_state=42)
    else:
        model_k = RandomForestClassifier(n_estimators=100, class_weight='balanced', random_state=42, n_jobs=-1)
    model_k.fit(Xtr_k, ytr_k_perm)
    cluster_models[k] = model_k
    
    pk = model_k.predict(Xte_k)

    # 同群內（公平）比較：global vs cluster-k
    acc_g_k = accuracy_score(yte_k, pred_global[idx_te])
    f1_g_k  = f1_score(yte_k, pred_global[idx_te], average='macro', zero_division=0)
    acc_k   = accuracy_score(yte_k, pk)
    f1_k    = f1_score(yte_k, pk, average='macro', zero_division=0)

    # 該群 train 的「unchanged」比例（對齊 index 後比較）
    idx_tr_k = Xtr_k.index
    unchanged_frac_k = (y_train.loc[idx_tr_k] == y_train_perm.loc[idx_tr_k]).mean()
    
    rows.append({'cluster': int(k), 'n_train': int(idx_tr.sum()), 'n_test': int(idx_te.sum()), 'unchanged_train': unchanged_frac_k,
                 'Global_Acc': acc_g_k, 'Global_MacroF1': f1_g_k,
                 'Cluster_Acc': acc_k, 'Cluster_MacroF1': f1_k,
                 'Delta_Acc': acc_k-acc_g_k, 'Delta_MacroF1': f1_k-f1_g_k})
    reports[k] = classification_report(yte_k, pk, zero_division=0)
    
# 6) 彙整
summary = pd.DataFrame(rows).sort_values('cluster').reset_index(drop=True)

print("=== Global model on all test ===")
print(f"Acc={acc_global:.3f}  MacroF1={f1_global:.3f}")
print("\n=== Cluster comparison ===")
print(summary)
```

```python
# === 新資料：路由 + 預測（含低信心退回 & top-2 對決） ===
import numpy as np
import pandas as pd

def route_and_predict(x_new_df: pd.DataFrame, method='corr', t=0.6):
    # x_new_df：一列 DataFrame，欄位順序= X_cols
    Xn = x_new_df[X_cols]
    router = router_corr if method=='corr' else router_euclid

    # 1) 路由機率與 top-2
    P = router.predict_proba(Xn)[0]
    labels = router.classes_
    top2 = np.argsort(P)[-2:][::-1]
    k1, k2 = int(labels[top2[0]]), int(labels[top2[1]]) if len(top2)>1 else (int(labels[top2[0]]), None)
    p1 = float(P[top2[0]])
    p2 = float(P[top2[1]]) if len(top2)>1 else 0.0

    # 2) 低信心退回全域
    if p1 < t or k1 not in cluster_models:
        y_hat = global_rfc.predict(Xn)[0]
        return {'route': 'global', 'y_pred': int(y_hat), 'p_router': p1, 'k1': k1, 'k2': k2}

    # 3) top-2 對決（若第二名存在且該群有模型）
    if k2 is not None and k2 in cluster_models:
        proba1 = float(np.max(cluster_models[k1].predict_proba(Xn)[0]))
        proba2 = float(np.max(cluster_models[k2].predict_proba(Xn)[0]))
        use_k = k1 if proba1 >= proba2 else k2
    else:
        use_k = k1

    y_hat = cluster_models[use_k].predict(Xn)[0]
    return {'route': int(use_k), 'y_pred': int(y_hat), 'p_router': p1}
```
